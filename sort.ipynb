```python
import pandas as pd
import os
import json
fileist=os.listdir("执业医师学科真题")
```

```python
data=pd.DataFrame()
for each in fileist:
    # each is json file
    with open("执业医师学科真题/"+each,"r",encoding="utf-8") as f:
        tmp=json.load(f)
        tmp['option']=str('<br/>'.join(tmp['option']))
        tmp['test'] = f"{tmp.get('main_test', '')}<br/>{tmp['test']}".strip() if tmp.get('main_test', '').strip() else tmp['test']
        #dict tmp删除main_test
        tmp.pop('main_test', None)
        tmp['tag']=str("蓝基因::执业医师::学科真题::"+tmp['cls']+"::"+tmp['unit'])
        data=pd.concat([data,pd.DataFrame([tmp])],axis=0)
```

```python
# 去重
data.drop_duplicates(subset=["test",'numb','tag'],inplace=True)
```

```python
# 保存data 为csv文件
data.to_csv("执业医师学科真题.tsv",index=False,encoding="utf-8",sep="\t")
```
